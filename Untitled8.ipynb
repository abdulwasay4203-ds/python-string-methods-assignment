```python
# Assignment 1
# Python String Methods Assignment

# Taking input from user
name = input("Enter your name: ")
sentence = input("Enter a sentence: ")
data = input("Enter numbers or text: ")

# 1. Case Conversion Methods
print("nCase Conversion Methods")
print("upper():", name.upper())
print("lower():", name.lower())
print("title():", name.title())
print("capitalize():", sentence.capitalize())
print("swapcase():", sentence.swapcase())
print("casefold():", name.casefold())

# 2. Space Removal Methods
print("Space Removal Methods")
print("strip():", data.strip())
print("lstrip():", data.lstrip())
print("rstrip():", data.rstrip())

# 3. Search and Replace Methods
print("Search and Replace Methods")
print("replace():", sentence.replace("Python", "Java"))
print("find():", sentence.find("a"))
print("index():", sentence.index(sentence[0]) if sentence else "Empty string")
print("count():", sentence.count("a"))
print("startswith():", sentence.startswith("I"))
print("endswith():", sentence.endswith("."))

# 4. Split and Join Methods
print("Split and Join Methods")
words = sentence.split()
print("split():", words)
print("join():", " | ".join(words))

# 5. Checking Methods
print("Checking Methods")
print("isalpha():", name.isalpha())
print("isdigit():", data.isdigit())
print("isalnum():", data.isalnum())
print("isspace():", data.isspace())
print("islower():", name.islower())
print("isupper():", name.isupper())
print("istitle():", name.istitle())

# 6. Alignment Methods
print("Alignment Methods")
print("center():", name.center(30, "-"))
print("ljust():", name.ljust(30, "-"))
print("rjust():", name.rjust(30, "-"))
print("zfill():", data.zfill(10))

# 7. Partition Methods
print("Partition Methods")
print("partition():", sentence.partition(" "))
print("rpartition():", sentence.rpartition(" "))
```
